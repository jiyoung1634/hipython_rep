```python
import requests
import folium
from pandas import json_normalize
```

```python
targetSite='https://www.starbucks.co.kr/store/getStore.do?r=8K3XF06R9Q'
req=requests.post(targetSite, data={
'ins_lat': 37.563398,
'ins_lng': 126.9863309,
'p_sido_cd': '01',
'p_gugun_cd': '',
'in_biz_cd': '',
'iend': 600,
'set_date': ''
})
starbucks=req.json()
print(type(starbucks))
```

```python
st_df = json_normalize(starbucks, 'list')
st_df
```

```python
st_df_map = st_df[['s_name','sido_code','sido_name','gugun_code','gugun_name',
'doro_address','lat','lot']]
st_df_map
```

```python
st_df_map.info()
```

```python
st_df_map['lat'] = st_df_map['lat'].astype(float)
st_df_map['lot'] = st_df_map['lot'].astype(float)
st_df_map.info()
```

```python
st_df_map.head()
```

```python
loc = [37.501087, 127.043069]
stMap = folium.Map(location=loc, zoom_start=18)
pop = folium.Popup('역삼아레나빌딩', max_width=100)
folium.Marker(location=loc, popup=pop).add_to(stMap)
stMap
```

```python
st_lat = []
for i in range(10):
    st_lat.append(st_df_map['lat'].loc[i])

st_lot = []
for u in range(10):
    st_lot.append(st_df_map['lot'].loc[i])
    
#loc = []
for m in range(10):
    loc = [st_lat[i], st_lot[i]]
    
    stMap = folium.Map(location=loc, zoom_start=18)
    pop = folium.Popup('역삼아레나빌딩', max_width=100)
    folium.Marker(location=loc, popup=pop).add_to(stMap)
```

```python
# 지도를 루프 밖에서 한 번만 생성
# 첫 번째 위치를 중심으로 지도 초기화
initial_loc = [st_df_map['lat'].loc[0], st_df_map['lot'].loc[0]]
stMap = folium.Map(location=initial_loc, zoom_start=18)

# 10개의 위치에 마커 추가
for i in range(10):
    # 'lat'과 'lot' 값을 가져와서 loc 리스트 생성
    lat_val = st_df_map['lat'].loc[i]
    lon_val = st_df_map['lot'].loc[i]
    loc = [lat_val, lon_val]
    
    # 팝업 내용 설정 (원하는 정보로 수정 가능)
    pop = folium.Popup(f'위치 {i+1}', max_width=100)
    
    # 마커를 지도에 추가
    folium.Marker(location=loc, popup=pop).add_to(stMap)

# 지도 저장 또는 출력
stMap
```
